```python
#演習1
%reset -f

W = 512

def ghost_ship_message(x1: int, y1: int, x2: int, y2: int) -> str:#関数を利用した。
    assert all(0 <= v <= 511 for v in (x1, y1, x2, y2)), "座標が範囲外です"

    east,  west  = (x2 - x1) % W, (x1 - x2) % W
    south, north = (y2 - y1) % W, (y1 - y2) % W #

    h_dir, h_dist = ("東", east)  if east  <= west  else ("西", west)
    v_dir, v_dist = ("南", south) if south <= north else ("北", north)

    if h_dist == 0 and v_dist == 0:
        return '💀「幽霊船にであえたようじゃな」'

    parts = [f"{d}に{n}" for d, n in ((h_dir, h_dist), (v_dir, v_dist)) if n != 0]
    return f'💀「{"、".join(parts)}進めば幽霊船にであえるだろう」'#ここまで20分経過した。

cases = [
    (20, 10, 30,  5),
    (10, 10,  7, 13),
    (100, 490, 100, 510),
    (200, 200, 490, 200),
    (50,  50,  50,  50),
]

for x1, y1, x2, y2 in cases:
    print(f'({x1},{y1}) -> ({x2},{y2})')
    print(ghost_ship_message(x1, y1, x2, y2))
    print() #ここまで30分経過した。

```

### 演習2の数式

現在位置 $(x_1, y_1)$ から目標 $(x_2, y_2)$ へ向かうベクトルは、

$$\vec{d} = (x_2 - x_1,\ y_2 - y_1)$$

### 2. 距離（ベクトルの大きさ）を求める

$$|\vec{d}| = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

### 3. 単位ベクトルを求める

方向はそのままに、大きさを1に正規化すると

$$\hat{d} = \frac{\vec{d}}{|\vec{d}|} = \left(\frac{x_2 - x_1}{|\vec{d}|},\ \frac{y_2 - y_1}{|\vec{d}|}\right)$$

### 4. t秒後の移動量を求める

速度が1単位距離/秒なので、$t$ 秒間の移動量は

$$\Delta = \hat{d} \times t$$

### 5. t秒後の位置を求める

ただし、移動量が距離を超える場合は目標位置で停止するので

$$
(x_3, y_3) = \begin{cases}
(x_1, y_1) & (|\vec{d}| = 0) \\
(x_2, y_2) & (t \geq |\vec{d}|) \\
\left(x_1 + \dfrac{x_2 - x_1}{|\vec{d}|} \cdot t,\ y_1 + \dfrac{y_2 - y_1}{|\vec{d}|} \cdot t\right) & (\text{それ以外})
\end{cases}
$$

---

### 逆算

$(x_1, y_1) = (0, 0)$、$(x_2, y_2) = (10, 10)$、$t = 8$ のとき：

$$|\vec{d}| = \sqrt{10^2 + 10^2} = \sqrt{200} \approx 14.14$$

$$t = 8 < 14.14 \text{ なので通常移動}$$

$$x_3 = 0 + \frac{10}{14.14} \times 8 \approx 5.66 \approx 5.7$$

$$y_3 = 0 + \frac{10}{14.14} \times 8 \approx 5.66 \approx 5.7$$

```python
#演習2
%reset -f
import math

x1, y1 = 0, 0
x2, y2 = 10, 10
t = 8
dx = x2 - x1
dy = y2 - y1
distance = math.sqrt(dx**2 + dy**2)

if distance == 0:
    x3 = x1
    y3 = y1
elif t >= distance:
    x3 = x2
    y3 = y2
else:
    x3 = x1 + (dx / distance) * t
    y3 = y1 + (dy / distance) * t

print(f'({x1:.1f}, {y1:.1f}) から ({x2:.1f}, {y2:.1f}) を目標に、')
print(f'速度1で移動するキャラの {t:.1f} 秒後の位置は ({x3:.1f}, {y3:.1f})')

cases = [
    (0, 0, 10, 10, 8)   ,                                        
    (0, 0, 10, 10, 12)  ,                                        
    (0, 0, 10, 10, 0)   ,                                        
    (5, 5, 5, 5, 8)     ,                                        
]   
```
