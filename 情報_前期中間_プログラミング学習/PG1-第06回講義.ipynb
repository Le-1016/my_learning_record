```python
#演習1
%reset -f

W = 512

def ghost_ship_message(x1: int, y1: int, x2: int, y2: int) -> str:#関数を利用した。
    assert all(0 <= v <= 511 for v in (x1, y1, x2, y2)), "座標が範囲外です"

    east,  west  = (x2 - x1) % W, (x1 - x2) % W
    south, north = (y2 - y1) % W, (y1 - y2) % W #

    h_dir, h_dist = ("東", east)  if east  <= west  else ("西", west)
    v_dir, v_dist = ("南", south) if south <= north else ("北", north)

    if h_dist == 0 and v_dist == 0:
        return '💀「幽霊船にであえたようじゃな」'

    parts = [f"{d}に{n}" for d, n in ((h_dir, h_dist), (v_dir, v_dist)) if n != 0]
    return f'💀「{"、".join(parts)}進めば幽霊船にであえるだろう」'#ここまで40分経過した。

cases = [
    (20, 10, 30,  5),
    (10, 10,  7, 13),
    (100, 490, 100, 510),
    (200, 200, 490, 200),
    (50,  50,  50,  50),
]

for x1, y1, x2, y2 in cases:
    print(f'({x1},{y1}) -> ({x2},{y2})')
    print(ghost_ship_message(x1, y1, x2, y2))
    print() #ここまで60分経過した。

```

```python
#演習2
%reset -f
import math

x1, y1 = 0, 0
x2, y2 = 10, 10
t = 8
dx = x2 - x1
dy = y2 - y1
distance = math.sqrt(dx**2 + dy**2)

if distance == 0:
    x3 = x1
    y3 = y1
elif t >= distance:
    x3 = x2
    y3 = y2
else:
    x3 = x1 + (dx / distance) * t
    y3 = y1 + (dy / distance) * t

print(f'({x1:.1f}, {y1:.1f}) から ({x2:.1f}, {y2:.1f}) を目標に、')
print(f'速度1で移動するキャラの {t:.1f} 秒後の位置は ({x3:.1f}, {y3:.1f})')

cases = [
    (0, 0, 10, 10, 8)   ,
    (0, 0, 10, 10, 12)  ,
    (0, 0, 10, 10, 0)   ,
    (5, 5, 5, 5, 8)     ,
]   ＃ここまで100分経過した。
```

```python
#演習3
%reset -f
%reset -f
import math
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib_inline.backend_inline import set_matplotlib_formats
set_matplotlib_formats('svg')
plt.rcParams['mathtext.fontset'] = 'cm'

x1, y1 =   1,   -4
x2, y2 =  -5, -1.5

assert x1 != x2 or y1 != y2, '点1と点2は異なる座標である必要があります（仕様違反）'

p1 = np.array([x1, y1])
p2 = np.array([x2, y2])
v = p2 - p1
v_unit = v / np.linalg.norm(v)
perp = np.array([-v_unit[1], v_unit[0]])
p1a = p1 + perp * 0.5
p1b = p1 - perp * 0.5
p2a = p2 + perp * 0.5
p2b = p2 - perp * 0.5

x1a, y1a = p1a
x1b, y1b = p1b
x2a, y2a = p2a
x2b, y2b = p2b

print(f'点1の座標は ({x1:.2f},{y1:.2f})')
print(f'  これを中点とする長さ1の垂直線の両端座標は ({x1a:.2f},{y1a:.2f}) と ({x1b:.2f},{y1b:.2f})')
print(f'点2の座標は ({x2:.2f},{y2:.2f})')
print(f'  これを中点とする長さ1の垂直線の両端座標は ({x2a:.2f},{y2a:.2f}) と ({x2b:.2f},{y2b:.2f})')#製作時間は60分程度でした。
```
